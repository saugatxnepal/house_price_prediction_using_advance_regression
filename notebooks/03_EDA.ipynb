```python
import pandas as pd              # Loads and works with tabular data (DataFrames)
import numpy as np               # Performs numerical calculations and works with arrays
import matplotlib.pyplot as plt  # Creates charts and plots
import seaborn as sns            # Creates statistical charts with simpler styling
```

```python
df=pd.read_csv('../data/processed/cleaned_data.csv')
df.head()
```

```python
plt.Figure(figsize=(10, 6))  # Set the figure size for the plot
sns.histplot(df["SalePrice"], kde=True)  # Create a histogram of the SalePrice column with a kernel density estimate
plt.xlabel("Sale Price")
plt.ylabel("Frequency")
plt.title("Distribution of Sale Prices")
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()["SalePrice"].sort_values(
    ascending=False
)

correlation.head(15)
```

```python
plt.figure(figsize=(12, 8))    # Set the figure size for the heatmap

numeric_df = df.select_dtypes(include=np.number)  # Select only numeric columns from the DataFrame

correlation_matrix = numeric_df.corr()  # Calculate the correlation matrix for the numeric columns

display(correlation_matrix)  # Display the correlation matrix

sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm")  # Create a heatmap of the correlation matrix with annotations
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="OverallQual",
    y="SalePrice",
    data=df
)   # Create a box plot of OverallQual vs SalePrice

plt.title("Overall Quality vs Sale Price")
plt.show()  # Display the plot
```
