```python
import pandas as pd # Loads and works with tabular data (DataFrames)
import numpy as np  # Provides support for large, multi-dimensional arrays and matrices
import joblib # Provides tools for pipelining Python jobs
from pathlib import Path

from sklearn.pipeline import Pipeline # Provides a way to streamline a lot of the routine processes in machine learning

from sklearn.linear_model import LinearRegression, Ridge, Lasso # provide implementations of linear models for regression tasks
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor   # provide implementations of ensemble methods for regression tasks
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score # provide functions to evaluate the performance of regression models

from sklearn.model_selection import RandomizedSearchCV, GridSearchCV # provide tools for hyperparameter tuning and model selection
```

```python
preprocessor = joblib.load("../models/preprocessor.pkl")

split_data = joblib.load("../models/train_test_split.pkl")

X_train = split_data["X_train"]
y_train = split_data["y_train"]
X_test = split_data["X_test"]
y_test = split_data["y_test"]

preprocessor
```

```python
lr_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])  # Linear Regression is a simple and widely used algorithm for regression tasks. It assumes a linear relationship between the input features and the target variable, and it estimates the coefficients of the linear equation that best fits the data.

lr_pipeline.fit(X_train, y_train)
```

```python
ridge_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', Ridge(
        alpha=1.0,  # Regularization strength; must be a positive float. Larger values specify stronger regularization.
        random_state=42  # Ensures reproducibility of results by setting a seed for the random number generator
    ))
])  # Ridge Regression is a type of linear regression that includes a regularization term to prevent overfitting. It adds a penalty to the loss function based on the magnitude of the coefficients, which helps to shrink them towards zero and reduce their impact on the model.
```

```python
rf_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(
        n_estimators=100,  # The number of trees in the forest. More trees can improve performance but also increase computation time.
        random_state=42,  # Ensures reproducibility of results by setting a seed for the random number generator
        n_jobs=-1  # Utilizes all available CPU cores for parallel processing, speeding up the training process
    ))
])  # Random Forest Regressor is an ensemble learning method that builds multiple decision trees and merges them together to get a more accurate and stable prediction. It is known for its robustness and ability to handle large datasets with higher dimensionality.
```

```python
gb_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', GradientBoostingRegressor(
        random_state=42,  # Ensures reproducibility of results by setting a seed for the random number generator
    ))
])  # Gradient Boosting Regressor is an ensemble learning method that builds a series of decision trees, where each tree corrects the errors of the previous one. It is known for its high predictive accuracy and ability to handle complex datasets.
```

### Create a function

```python
def evaluate_model(model, X_test, y_test):
    predictions = model.predict(X_test)
    mae = mean_absolute_error(y_test, predictions)
    mse = mean_squared_error(y_test, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, predictions)

    return mae, mse, rmse, r2
```

### Create Comparison Table

```python
results = []

models = {
    "Linear Regression": lr_pipeline,
    "Ridge Regression": ridge_pipeline,
    "Random Forest Regressor": rf_pipeline,
    "Gradient Boosting Regressor": gb_pipeline
}

for model_name, model in models.items():
    model.fit(X_train, y_train)


    mae, mse, rmse, r2 = evaluate_model(
        model, X_test, y_test
    )

    results.append({
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })

    results_df = pd.DataFrame(results)

    results_df.sort_values(
        "RMSE"
    )
```

```python
gb_param_grid = {
    "model__n_estimators": [100, 200, 300, 500],
    "model__learning_rate": [0.01, 0.03, 0.05, 0.1],
    "model__max_depth": [2, 3, 4, 5],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4],
    "model__subsample": [0.8, 0.9, 1.0]
}   # This is from randomized search documentation. It defines a grid of hyperparameters for the Gradient Boosting Regressor model. Each key corresponds to a hyperparameter, and the values are lists of possible values to try during hyperparameter tuning.
```

```python
gb_search = RandomizedSearchCV(
    gb_pipeline,
    param_distributions=gb_param_grid,
    n_iter=30,  # Number of parameter settings that are sampled. n_iter trades off runtime vs quality of the solution. More iterations can lead to better results but will take longer to compute.
    scoring='neg_root_mean_squared_error',  # The metric used to evaluate the performance of the model during hyperparameter tuning. In this case, we are using the negative root mean squared error (RMSE) as the scoring metric. The negative sign is used because scikit-learn's RandomizedSearchCV maximizes the score, and we want to minimize RMSE.
    cv=5,  # Number of folds in cross-validation. The dataset is split into 'cv' number of folds, and the model is trained and evaluated 'cv' times, each time using a different fold as the validation set and the remaining folds as the training set.
    n_jobs=-1,  # Utilizes all available CPU cores for parallel processing, speeding up the hyperparameter search process.
    verbose=1,  # Controls the verbosity of the output. A higher value provides more detailed information about the progress of the search, which can be useful for monitoring long-running processes.
    random_state=42  # Ensures reproducibility of results by setting a seed for the random number generator
)
```

```python
gb_search.fit(X_train, y_train)
```

```python
gb_search.best_params_
```

```python
gb_search.best_score_
```

```python
best_cv_rmse = -gb_search.best_score_  # The best score from the RandomizedSearchCV is negative because we used 'neg_root_mean_squared_error' as the scoring metric. We negate it to get the actual RMSE value.

print(f"Best CV RMSE: {best_cv_rmse:.4f}")  # This line prints the best cross-validated RMSE obtained from the hyperparameter tuning process. The value is formatted to four decimal places for better readability.
```

```python
best_gb = gb_search.best_estimator_  # This line retrieves the best estimator (model) found during the hyperparameter tuning process. The best estimator is the model that achieved the lowest RMSE on the cross-validation folds, and it is ready to be used for making predictions on new data.
best_gb
```

```python
y_pred = best_gb.predict(X_test)
y_pred
```
