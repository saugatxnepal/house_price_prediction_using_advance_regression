```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
split_data = joblib.load('../models/train_test_split.pkl')  # Load the train-test split data
# split_data.keys()

y_test = split_data["y_test"]   # Load the test target variable

y_pred = joblib.load('../models/y_pred.pkl')  # Load the predicted values from the model
```

```python
mae = mean_absolute_error(y_test, y_pred)   # Calculate Mean Absolute Error

rmse = np.sqrt(mean_squared_error(y_test, y_pred))  # Calculate Root Mean Squared Error

r2 = r2_score(y_test, y_pred)  # Calculate R-squared score

print(f"Mean Absolute Error (MAE): {mae}")
print(f"Root Mean Squared Error (RMSE): {rmse}")
print(f"R-squared Score: {r2}")
```

```python
comparision = pd.DataFrame({
    "Metrices": ["Mean Absolute Error (MAE)", "Root Mean Squared Error (RMSE)", "R-squared Score"],
    "Before Hyperparameter Tuning": [16336.978872, 26956.543216, 0.905264],
    "After Hyperparameter Tuning": [mae, rmse, r2]
})  #   create a DataFrame to compare the metrics before and after hyperparameter tuning

comparision
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred, alpha=0.5)

plt.xlabel('Actual Prices')
plt.ylabel('Predicted Prices')
plt.title('Actual vs Predicted Prices')

plt.show()
```

```python
residuals = y_test - y_pred  # Calculate residuals 
```

```python
plt.Figure(figsize=(10, 6))

sns.histplot(residuals, kde=True)

plt.xlabel('Residuals')
plt.ylabel('Frequency')

plt.show()
```
